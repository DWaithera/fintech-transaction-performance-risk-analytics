# Initial Data Profiling Summary

## Dataset Overview

The PaySim dataset contains **6,362,620 transactions across 11 columns**.

The dataset contains transaction-level information covering:
- Transaction timing (`step`)
- Transaction type
- Transaction amount
- Origin and destination accounts
- Origin and destination balances
- Fraud labels
- Fraud detection flags

## Initial Data Quality Findings

| Check | Finding | Status |
|---|---:|---|
| Rows | 6,362,620 | ✅ |
| Columns | 11 | ✅ |
| Missing values | 0 | ✅ |
| Duplicate rows | 0 | ✅ |
| Negative transaction amounts | 0 | ✅ |
| Negative balances | 0 | ✅ |
| Zero-value transactions | 16 | ⚠️ Investigate |
| Observed steps | 1–743 | ✅ |
| Missing steps within observed range | 0 | ✅ |

## Transaction Mix

The dataset contains five transaction types:

- CASH_OUT — 35.17%
- PAYMENT — 33.81%
- CASH_IN — 21.99%
- TRANSFER — 8.38%
- DEBIT — 0.65%

CASH_OUT and PAYMENT together account for approximately **68.98% of all transactions**.

## Fraud Profile

There are **8,213 transactions labeled as fraudulent**, representing approximately **0.1291%** of all transactions.

The simulated fraud-detection flag is present on only **16 transactions**.

Fraud labels are observed in:
- CASH_OUT — 0.1840%
- TRANSFER — 0.7688%

No fraud labels are observed in CASH_IN, DEBIT, or PAYMENT transactions in this dataset.

These are descriptive observations only. Further validation is required before interpreting them as business risk patterns.

## Data Quality Observations Requiring Further Investigation

### 1. Zero-value transactions

There are 16 transactions with an amount of zero. All 16 are CASH_OUT transactions.

These transactions have not been removed or modified. Their behavior will be investigated during the data-quality validation stage.

### 2. Balance reconciliation

Initial balance reconciliation shows that origin and destination balances do not consistently reconcile to the transaction amount.

This is an important validation finding because balance movements are central to understanding transaction behavior and financial integrity.

The discrepancies will **not be corrected during initial profiling**.

They will be investigated during the next stage by:
- transaction type
- fraud status
- account behavior
- balance patterns
- transaction-level examples

## Analytical Principle

> **Profile first. Validate second. Transform third. Analyze fourth. Visualize last.**

Initial profiling identifies patterns and potential data-quality issues. It does not establish causality or justify modifying the source data.

## Conclusion

The dataset is structurally usable for further analysis, with no missing values, duplicate rows, negative transaction amounts, or negative balances identified during initial profiling.

However, several areas require deeper validation before analytical modeling:

1. Balance reconciliation
2. Zero-value transactions
3. Fraud-label and fraud-flag behavior
4. Transaction-type-specific data behavior
5. Account-level transaction patterns

These issues will be investigated before transformation and KPI development.
